# Exploración de factores asociados a obesidad

## Objetivo
Explorar un conjunto de datos con características físicas, hábitos y nivel de obesidad para describir su estructura, resumir variables relevantes y analizar gráficamente una subpoblación.

## Datos
El conjunto contiene 2,111 registros y 10 variables. Incluye variables numéricas como edad, estatura, peso, consumo de vegetales y consumo de agua, además de variables categóricas relacionadas con hábitos y nivel de obesidad.

## Metodología
Se utiliza Python con pandas y matplotlib para cargar los datos, revisar estructura y valores faltantes, generar estadísticas descriptivas, estudiar la distribución de `NivelDeObesidad` y comparar el peso entre categorías dentro de la subpoblación masculina.


```python
import pandas as pd
import matplotlib.pyplot as plt

print("Librerías cargadas correctamente")
```

```python
try:
    datos = pd.read_csv("A1.1 Obesidad.csv")
    print("DEBUG: Archivo cargado correctamente")
    print("Filas y columnas:", datos.shape)
except Exception as e:
    print("Error al cargar el archivo:", e)
```

```python
try:
    print("DEBUG: Mostrando las primeras 5 filas")
    print(datos.head())
except Exception as e:
    print("Error al mostrar los datos:", e)
```

```python
try:
    print("DEBUG: Revisando la estructura de los datos")

    print("\nTipos de datos:")
    print(datos.dtypes)

    print("\nValores faltantes por columna:")
    print(datos.isnull().sum())

except Exception as e:
    print("Error al revisar la estructura de los datos:", e)
```

```python
try:
    print("DEBUG: Generando resumen estadístico")
    print(datos.describe())
except Exception as e:
    print("Error al generar el resumen estadístico:", e)
```

```python
try:
    print("DEBUG: Analizando la variable NivelDeObesidad")

    conteo_obesidad = datos["NivelDeObesidad"].value_counts()

    print("\nCantidad de personas por nivel de obesidad:")
    print(conteo_obesidad)

except Exception as e:
    print("Error al analizar NivelDeObesidad:", e)
```

```python
try:
    print("DEBUG: Generando gráfica de NivelDeObesidad")

    conteo_obesidad.plot(kind="bar")

    plt.title("Distribución de los niveles de obesidad")
    plt.xlabel("Nivel de obesidad")
    plt.ylabel("Cantidad de personas")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

except Exception as e:
    print("Error al generar la gráfica:", e)
```

```python
try:
    print("DEBUG: Filtrando subpoblación masculina")

    hombres = datos[datos["Sexo"] == "Male"]

    print("Cantidad de hombres:", hombres.shape[0])

    hombres.boxplot(column="Peso", by="NivelDeObesidad", figsize=(10, 6))

    plt.title("Peso según nivel de obesidad en hombres")
    plt.suptitle("")
    plt.xlabel("Nivel de obesidad")
    plt.ylabel("Peso")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

except Exception as e:
    print("Error al analizar la subpoblación:", e)
```

```python
try:
    print("DEBUG: Contando hombres por nivel de obesidad")

    conteo_hombres = hombres["NivelDeObesidad"].value_counts()

    print("\nCantidad de hombres por nivel de obesidad:")
    print(conteo_hombres)

except Exception as e:
    print("Error al contar los niveles de obesidad:", e)
```

## Conclusiones
El conjunto no presenta valores faltantes y permite observar diferencias claras en el peso entre categorías de obesidad. En la subpoblación masculina se aprecia una tendencia general de aumento del peso conforme aumenta el nivel de obesidad. Algunas categorías deben interpretarse con cautela cuando tienen muy pocos registros.
